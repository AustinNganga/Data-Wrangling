Question 5: Create fatality_locations dataframe

```python
# Select only the required columns from the raw dataset
# .copy() ensures this is a new independent dataframe, not a view of df
fatality_locations = df[['Date', 'Location', 'Aboard', 'Fatalities']].copy()

# Display the new dataframe
fatality_locations
```

Question 6: Date with highest number of fatalities

```python
# idxmax() finds the index label of the row with the maximum 'Fatalities' value
# .loc[] then retrieves that entire row
max_row = fatality_locations.loc[fatality_locations['Fatalities'].idxmax()]

# Print the date and fatality count for that row
print("Date:", max_row['Date'])
print("Fatalities:", max_row['Fatalities'])
```

Question 7: Aboard vs Fatalities, and crashes with no fatalities

```python
# Filter the dataframe to keep only rows where Fatalities is exactly 0
no_fatalities = fatality_locations[fatality_locations['Fatalities'] == 0]

# Count how many crashes had zero fatalities
print("Crashes with no fatalities:", len(no_fatalities))

# Total number of crashes in the dataframe, for context/comparison
print("Total crashes:", len(fatality_locations))
```

```python
# Question 8: Split the 'Location' column into 'Region' and 'State_Country'
# str.split(',', n=1, expand=True) splits on only the FIRST comma,
# which correctly handles locations with extra commas (e.g. "City, County, State")
split_location = fatality_locations['Location'].str.split(',', n=1, expand=True)

# Assign the two parts back as new columns, stripping leading/trailing whitespace
fatality_locations['Region'] = split_location[0].str.strip()
fatality_locations['State_Country'] = split_location[1].str.strip()

# Display the result
fatality_locations
```

```python
# Question 9: Sort by Fatalities descending and keep the top 100
top_100_fatalities = fatality_locations.sort_values(by='Fatalities', ascending=False).head(100)

top_100_fatalities
```

```python
# Question 10: Pie chart of fatalities distribution by country/state (top 25)
import matplotlib.pyplot as plt
import seaborn as sns

# Group the top 100 crashes by State_Country and sum fatalities per group
fatalities_by_country = top_100_fatalities.groupby('State_Country')['Fatalities'].sum()

# Take the top 25 countries/states by total fatalities
top_25_countries = fatalities_by_country.sort_values(ascending=False).head(25)

# Use a seaborn color palette for the pie slices
colors = sns.color_palette('pastel', len(top_25_countries))

plt.figure(figsize=(10, 10))
plt.pie(top_25_countries, labels=top_25_countries.index, colors=colors, autopct='%1.1f%%')
plt.title('Distribution of Fatalities by Country/State (Top 25)')
plt.show()
```
